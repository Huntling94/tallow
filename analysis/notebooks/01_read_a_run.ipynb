# Read a run

Loads one telemetry file written by the headless runner and shows what is in it. Make one first, from the repository root:

```bash
godot --headless --path game -s res://tools/headless_runner.gd -- --seed 42 --seasons 6 --out ../telemetry/run_42.jsonl
```

If that file is missing, the notebook falls back to the small sample in `fixtures/`. The herd sizes are placeholders until the real herd rules arrive in M1.

````python
from pathlib import Path

import matplotlib.pyplot as plt

from tallow_analysis.telemetry import load_run

run_file = Path("../../telemetry/run_42.jsonl")
if not run_file.exists():
    run_file = Path("../fixtures/run_complete.jsonl")

run = load_run(run_file)
print(f"File:     {run_file}")
print(f"Run id:   {run.run_id}")
print(f"Seed:     {run.header['seed']}")
print(f"Rules:    {run.header['rules_version']}")
print(f"Settings: {run.header['settings']}")
print(f"Events:   {len(run.events)}, complete: {run.complete}")
````

````python
arrivals = [event for event in run.events if event["type"] == "herd_arrived"]
seasons = [event["season"] for event in arrivals]
sizes = [event["herd_size"] for event in arrivals]

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.bar(seasons, sizes, color="#b0703c")
ax.set_xlabel("Season")
ax.set_ylabel("Herd size (placeholder)")
ax.set_title(f"Herd size by season, seed {run.header['seed']}")
ax.set_xticks(seasons)
ax.spines[["top", "right"]].set_visible(False)
plt.show()
````
